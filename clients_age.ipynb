```python
age_lst = [23, 45, 18, 80, 22, 25, 34, 37, 41, 88, 19, 20, 18, 45, 41, 39, 31, 32, 22, 28, 37]
```

```python
avg_age = sum(age_lst)/len(age_lst)
print(f"Средний возраст: {round(avg_age, 2)} лет")
```

```python
# Фильтрация данных
sum_age = 0
count_age = 0
for age in age_lst:
    if age < 50:
        sum_age += age
        count_age += 1
avg_age = sum_age/count_age
print(f"Средний возраст посетителей младше 50 лет: {round(avg_age, 2)} лет")        
```

```python
# Функция с порогом
def calc_age(lst, thresh):
    sum_age = 0
    count_age = 0
    for age in lst:
        #проверка числового типа у значение массива
        try:
            if age < thresh:
                sum_age += age
                count_age += 1
        except TypeError:
            print(f"{age} это нечисловое значение")
    # проверка количества подходящих по значению элементов 
    try:
        avg_age = sum_age/count_age
        return avg_age
    except ZeroDivisionError:
        print(f"список должен содержать хотя-бы одно значение меньше {thresh}")

max_age = 50
try:
    a = calc_age(age_lst, max_age)
    print(f"Средний возраст посетителей младше {max_age} лет: {round(a, 2)} лет")
except TypeError:
    print("в работе функции было вызвано исключение")
```

```python
import numpy as np
```

```python
# Автоматическое исключение выбросов
q1 = np.percentile(age_lst, 25)
q3 = np.percentile(age_lst, 75)
iqr = q3-q1
lower_bound = q1-1.5*iqr
upper_bound = q3+1.5*iqr
sum_age = 0
count_age = 0
for age in age_lst:
    if (age < upper_bound) and (age > lower_bound):
        sum_age += age
        count_age += 1
avg_age = sum_age/count_age
print(f"Средний возраст посетителей за исключений выбросов: {round(avg_age, 2)} лет")        
```

```python
# Универсальная функция для выбросов
def calc_age(lst):
    try:
        q1 = np.percentile(lst, 25)
        q3 = np.percentile(lst, 75)
        iqr = q3-q1
        lower_bound = q1-1.5*iqr
        upper_bound = q3+1.5*iqr
        sum_age = 0
        count_age = 0
        for age in lst:
            if (age < upper_bound) and (age > lower_bound):
                sum_age += age
                count_age += 1
        avg_age = sum_age/count_age
        return avg_age
    except:
        print('Список не должен содержать нечиcловых значений или быть пустым')
try:
    a = calc_age(age_lst)
    print(f"Средний возраст посетителей за исключений выбросов: {round(a, 2)} лет")   
except TypeError:
    print("в работе функции было вызвано исключение")
```
